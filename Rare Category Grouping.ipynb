```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('zomato.csv')
```

```python
df.head()
```

```python
df.info()
```

```python
df[['rest_type', 'cuisines']].head()
```

```python
df[['rest_type', 'cuisines']].isnull().sum()
```

```python
df['rest_type'] = df['rest_type'].fillna('Unknown')

df['cuisines'] = df['cuisines'].fillna('Unknown')
```

```python
print("Rest Type:", df['rest_type'].nunique())

print("Cuisines:", df['cuisines'].nunique())
```

```python
df['rest_type'].value_counts()
```

```python
df['cuisines'].value_counts()
```

```python
threshold = 100
```

```python
rest_counts = df['rest_type'].value_counts()

df['rest_type'] = df['rest_type'].apply(
    lambda x: x if rest_counts[x] >= threshold else 'Other'
)
```

```python
cuisine_counts = df['cuisines'].value_counts()

df['cuisines'] = df['cuisines'].apply(
    lambda x: x if cuisine_counts[x] >= threshold else 'Other'
)
```

```python
df['rest_type'].value_counts()
```

```python
df['cuisines'].value_counts()
```

```python
print("Final Rest Type Categories:", df['rest_type'].nunique())

print("Final Cuisines Categories:", df['cuisines'].nunique())
```

```python
df[['rest_type', 'cuisines']].isnull().sum()
```

```python
df.to_csv('zomato_rare_categories_grouped.csv', index=False)
```
